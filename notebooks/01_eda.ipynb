Run `python -m src.train` first if you want saved artifacts; this notebook explores the discovered dataset interactively.

```python
from src.data_loader import load_reviews, dataset_overview
from src.eda import save_eda_figures

reviews, schema = load_reviews()
schema, dataset_overview(reviews)
```

```python
reviews.info()
reviews.isna().sum().sort_values(ascending=False).head(15)
```

```python
save_eda_figures(reviews, schema['hotel_column'])
reviews.assign(class_name=reviews.label.map({0: 'Truthful', 1: 'Deceptive'})).groupby('class_name').size()
```

```python
# Optional columns are only analyzed when they exist.
for column in (schema['polarity_column'], schema['hotel_column']):
    if column:
        display(reviews.groupby(['label', column]).size().rename('count').reset_index().head(20))
```
